# Augmentierungsvergleich
Gefärbt → Ungefärbt angleichen für Domain Transfer Training

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import torchvision.transforms.functional as TF
from pathlib import Path
import cv2
import random
```

## 1. Bilder laden

```python
# ── Pfade anpassen ────────────────────────────────────────────────
cwd = Path.cwd().parent.parent
path_ungefaerbt = Path(cwd/'data/training/train/img_uncol/img/PE-2024-01126-M_00_s0021_PM_Complete_Transmittance_Stitched_Flat_v004_384_3328.png')
path_gefaerbt   = Path(cwd/'data/training/train/img/PE-2025-01953-M_00_s0070_PM_Complete_Transmittance_Stitched_Flat_v000_512_4352.png')
path_maske      = Path(cwd/'data/training/train/mask/PE-2025-01953-M_00_s0070_PM_Complete_Transmittance_Stitched_Flat_v000_512_4352.png')


to_tensor = lambda img: torch.tensor(
    np.array(img), dtype=torch.float32).unsqueeze(0) / 255.0

t_g = to_tensor(Image.open(path_gefaerbt).convert('L'))
t_u = to_tensor(Image.open(path_ungefaerbt).convert('L'))
t_m = (to_tensor(Image.open(path_maske).convert('L')) > 0.5).float()

print(f'Gefärbt:   min={t_g.min():.2f}  max={t_g.max():.2f}  mean={t_g.mean():.2f}')
print(f'Ungefärbt: min={t_u.min():.2f}  max={t_u.max():.2f}  mean={t_u.mean():.2f}')
```

## 2. Ausgangszustand

```python
def overlay_mask(ax, mask):
    o = np.zeros((*mask.squeeze().shape, 4))
    o[..., 0] = 1.0
    o[..., 3] = mask.squeeze().numpy() * 0.45
    ax.imshow(o)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
axes[0].imshow(t_g.squeeze().numpy(), cmap='gray', vmin=0, vmax=1)
axes[0].set_title('Gefärbt (Trainings-Input)')
overlay_mask(axes[0], t_m)
axes[1].imshow(t_g.squeeze().numpy(), cmap='gray', vmin=0, vmax=1)
axes[1].set_title('Gefärbt + Maske')
overlay_mask(axes[1], t_m)
axes[2].imshow(t_u.squeeze().numpy(), cmap='gray', vmin=0, vmax=1)
axes[2].set_title('Ungefärbt (Ziel-Domäne, keine Maske)')
for ax in axes: ax.axis('off')
plt.tight_layout()
plt.show()
```

## 3. Histogramm-Vergleich
Zeigt die Lücke die Augmentierung schließen soll.

```python
fig, ax = plt.subplots(figsize=(8, 3))
ax.hist(t_g.numpy().flatten(), bins=100, color='salmon',    alpha=0.7, label='Gefärbt')
ax.hist(t_u.numpy().flatten(), bins=100, color='steelblue', alpha=0.6, label='Ungefärbt (Ziel)')
ax.legend()
ax.set_title('Intensitätsverteilung: Gefärbt vs. Ungefärbt')
plt.tight_layout()
plt.show()
print(f'Mittlere Intensität – Gefärbt: {t_g.mean():.3f}  |  Ungefärbt: {t_u.mean():.3f}')
print(f'Standardabweichung  – Gefärbt: {t_g.std():.3f}   |  Ungefärbt: {t_u.std():.3f}')
```

## 4. Augmentierungen definieren
Alle in Richtung ungefärbt: Kontrast ↓, Helligkeit ↑, Blur, Histogram Matching.

```python
def aug_contrast_reduce(img, factor=0.5):
    """Kontrast reduzieren – gefärbt hat mehr Kontrast als ungefärbt"""
    return TF.adjust_contrast(img, factor)

def aug_brightness_increase(img, factor=1.4):
    """Helligkeit erhöhen – ungefärbt ist blasser"""
    return TF.adjust_brightness(img, factor)

def aug_gamma(img, gamma=0.7):
    """Gamma < 1 macht Bild heller, hebt Mitteltöne an"""
    return TF.adjust_gamma(img, gamma)

def aug_blur(img, sigma=1.5):
    """Weichzeichnen – ungefärbt hat weniger scharfe Gefäßränder"""
    arr = (img.squeeze().numpy() * 255).astype(np.uint8)
    blurred = cv2.GaussianBlur(arr, (0, 0), sigma)
    return torch.tensor(blurred, dtype=torch.float32).unsqueeze(0) / 255.0

def aug_histogram_match(img, reference):
    """Histogram Matching – passt Intensitätsverteilung direkt an Ziel an"""
    src = (img.squeeze().numpy() * 255).astype(np.uint8)
    ref = (reference.squeeze().numpy() * 255).astype(np.uint8)
    src_cdf = np.histogram(src.flatten(), 256, [0,256])[0].cumsum()
    ref_cdf = np.histogram(ref.flatten(), 256, [0,256])[0].cumsum()
    src_cdf = src_cdf / src_cdf[-1]
    ref_cdf = ref_cdf / ref_cdf[-1]
    lut = np.zeros(256, dtype=np.uint8)
    j = 0
    for i in range(256):
        while j < 255 and ref_cdf[j] < src_cdf[i]: j += 1
        lut[i] = j
    matched = lut[src]
    return torch.tensor(matched, dtype=torch.float32).unsqueeze(0) / 255.0

def aug_combined(img):
    """Kombination: Kontrast↓ + Helligkeit↑ + leichter Blur"""
    img = aug_contrast_reduce(img, factor=0.6)
    img = aug_brightness_increase(img, factor=1.3)
    img = aug_blur(img, sigma=1.0)
    return img.clamp(0, 1)

print('Augmentierungen definiert.')
```

## 5. Visueller Vergleich
Gefärbtes Bild mit verschiedenen Augmentierungen – Ziel ganz rechts.

```python
augmentierungen = {
    'Original (gefärbt)':   t_g,
    'Kontrast ↓ (0.5)':    aug_contrast_reduce(t_g, 0.5),
    'Helligkeit ↑ (1.4)':  aug_brightness_increase(t_g, 1.4),
    'Gamma ↓ (0.7)':       aug_gamma(t_g, 0.7),
    'Blur (σ=1.5)':        aug_blur(t_g, 1.5),
    'Hist. Matching':       aug_histogram_match(t_g, t_u),
    'Kombiniert':           aug_combined(t_g),
}

n = len(augmentierungen) + 1
fig, axes = plt.subplots(1, n, figsize=(4*n, 4))
for ax, (title, img) in zip(axes, augmentierungen.items()):
    ax.imshow(img.squeeze().numpy(), cmap='gray', vmin=0, vmax=1)
    ax.set_title(title, fontsize=8)
    ax.axis('off')

axes[-1].imshow(t_u.squeeze().numpy(), cmap='gray', vmin=0, vmax=1)
axes[-1].set_title('ZIEL: Ungefärbt', fontsize=8, color='steelblue', fontweight='bold')
axes[-1].axis('off')
for spine in axes[-1].spines.values():
    spine.set_edgecolor('steelblue'); spine.set_linewidth(3)

plt.suptitle('Augmentierungen des gefärbten Bildes → Richtung ungefärbt', fontsize=11)
plt.tight_layout()
plt.show()
```

## 6. MSE zum Ziel (quantitativ)

```python
mses = {name: ((img - t_u)**2).mean().item() for name, img in augmentierungen.items()}
beste_name = min((k for k in mses if k != 'Original (gefärbt)'), key=lambda k: mses[k])

print(f"{'Augmentierung':<25} {'MSE zum Ungefärbten':>20}")
print('-' * 47)
for name, mse in mses.items():
    marker = '  ◄ beste' if name == beste_name else ''
    print(f"{name:<25} {mse:>20.4f}{marker}")
```

## 7. Training-Preview: Augmentiertes Bild + originale Maske

```python
# Beste Augmentierung aus Zelle 6 automatisch übernehmen
beste = augmentierungen[beste_name]

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for ax, (img, title) in zip(axes, [
    (t_g,   'Original (gefärbt)'),
    (beste, f'Augmentiert ({beste_name})'),
    (t_u,   'Ziel: Ungefärbt'),
    (beste, 'Augmentiert + Maske'),
]):
    ax.imshow(img.squeeze().numpy(), cmap='gray', vmin=0, vmax=1)
    ax.set_title(title, fontsize=9)
    ax.axis('off')
overlay_mask(axes[-1], t_m)

plt.suptitle('Training Sample: augmentiertes Bild + originale Maske bleibt unverändert', fontsize=10)
plt.tight_layout()
plt.show()
```

## 8. Zufälliges Sampling – so sieht ein echter Trainings-Batch aus

```python
def random_aug(img):
    """Zufällige Augmentierung mit gesampelten Parametern für Training"""
    img = aug_contrast_reduce(img,     factor=random.uniform(0.4, 0.8))
    img = aug_brightness_increase(img, factor=random.uniform(1.1, 1.6))
    if random.random() > 0.5:
        img = aug_blur(img, sigma=random.uniform(0.5, 2.0))
    if random.random() > 0.5:
        img = aug_gamma(img, gamma=random.uniform(0.6, 0.9))
    return img.clamp(0, 1)

fig, axes = plt.subplots(2, 5, figsize=(20, 8))
for ax in axes.flatten():
    sample = random_aug(t_g)
    ax.imshow(sample.squeeze().numpy(), cmap='gray', vmin=0, vmax=1)
    overlay_mask(ax, t_m)
    ax.axis('off')

plt.suptitle('10 zufällige Trainings-Samples (gefärbt + random aug + Maske)', fontsize=11)
plt.tight_layout()
plt.show()
```
